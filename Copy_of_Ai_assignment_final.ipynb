```python
import networkx as nx
import matplotlib.pyplot as plt
from collections import deque

graph = {
    '2': ['4', '0'],
    '4': ['3', '1'],
    '0': ['5'],
    '3': ['5'],
    '1': ['5'],
    '5': []
}

def BFS(start, target):

    queue = deque([[start]])
    visited = []
    steps = []

    while queue:

        path = queue.popleft()
        node = path[-1]

        if node in visited:
            continue

        visited.append(node)
        steps.append((node, visited.copy()))

        if node == target:
            print("Path:", path)
            print("Visited Nodes:", visited)
            break

        for neighbor in graph[node]:
            if neighbor not in visited:
                queue.append(path + [neighbor])

    G = nx.DiGraph()

    for node, neighbors in graph.items():
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

    pos = {
        '2': (0, 3),
        '4': (-1, 2),
        '0': (1, 2),
        '3': (-1.5, 1),
        '1': (-0.5, 1),
        '5': (1, 1)
    }

    for node, done in steps:

        plt.figure(figsize=(8, 5))

        colors = [
            'orange' if n == node
            else 'lightgreen' if n in done
            else 'skyblue'
            for n in G.nodes()
        ]

        nx.draw(
            G, pos,
            with_labels=True,
            node_color=colors,
            node_size=2200,
            font_size=15,
            font_weight='bold',
            arrows=True
        )

        plt.title(
            f"BFS - ID 242031029 | Sohan | Visiting: {node}"
        )

        plt.show()


BFS('2', '5')
```

```python
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    '2': ['4', '0'],
    '4': ['3', '1'],
    '0': ['5'],
    '3': ['5'],
    '1': ['5'],
    '5': []
}

def IDS(start, target):

    G = nx.DiGraph()

    for node, neighbors in graph.items():
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

    pos = {
        '2': (0, 3),
        '4': (-1, 2),
        '0': (1, 2),
        '3': (-1.5, 1),
        '1': (-0.5, 1),
        '5': (1, 1)
    }

    def DLS(node, target, limit, path, visited, steps):

        visited.append(node)
        path.append(node)

        steps.append((node, visited.copy()))

        if node == target:
            return True

        if limit == 0:
            path.pop()
            return False

        for neighbor in graph[node]:

            if neighbor not in path:

                if DLS(
                    neighbor,
                    target,
                    limit - 1,
                    path,
                    visited,
                    steps
                ):
                    return True

        path.pop()
        return False

    for depth in range(len(graph)):

        print("\nDepth Limit:", depth)

        path = []
        visited = []
        steps = []

        found = DLS(
            start,
            target,
            depth,
            path,
            visited,
            steps
        )

        print("Visited Nodes:", visited)

        if found:
            print("Path:", path)

        for node, done in steps:

            plt.figure(figsize=(8, 5))

            colors = [
                'orange' if n == node
                else 'lightgreen' if n in done
                else 'skyblue'
                for n in G.nodes()
            ]

            nx.draw(
                G, pos,
                with_labels=True,
                node_color=colors,
                node_size=2200,
                font_size=15,
                font_weight='bold',
                arrows=True
            )

            plt.title(
                f"IDS - ID 242031029 | Sohan | Visiting: {node} | Limit: {depth}"
            )

            plt.show()

        if found:
            break


IDS('2', '5')
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import heapq

graph = {
    '2': [('4', 2), ('0', 4)],
    '4': [('3', 1), ('1', 5)],
    '0': [('5', 2)],
    '3': [('5', 3)],
    '1': [('5', 2)],
    '5': []
}

def UCS(start, target):

    pq = [(0, start, [start])]
    visited = []
    steps = []

    while pq:

        cost, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        visited.append(node)
        steps.append((node, visited.copy(), cost))

        if node == target:
            print("Path:", path)
            print("Total Cost:", cost)
            break

        for neighbor, weight in graph[node]:

            if neighbor not in visited:

                heapq.heappush(
                    pq,
                    (cost + weight, neighbor, path + [neighbor])
                )

    G = nx.DiGraph()

    for node, neighbors in graph.items():
        for neighbor, weight in neighbors:
            G.add_edge(node, neighbor, weight=weight)

    pos = {
        '2': (0, 3),
        '4': (-1, 2),
        '0': (1, 2),
        '3': (-1.5, 1),
        '1': (-0.5, 1),
        '5': (1, 1)
    }

    for node, done, cost in steps:

        plt.figure(figsize=(8, 5))

        colors = [
            'orange' if n == node
            else 'lightgreen' if n in done
            else 'skyblue'
            for n in G.nodes()
        ]

        nx.draw(
            G, pos,
            with_labels=True,
            node_color=colors,
            node_size=2200,
            font_size=15,
            font_weight='bold',
            arrows=True
        )

        nx.draw_networkx_edge_labels(
            G, pos,
            edge_labels=nx.get_edge_attributes(G, 'weight')
        )

        plt.title(
            f"UCS - ID 242031029 | Sohan | Visiting: {node} | Cost: {cost}"
        )

        plt.show()


UCS('2', '5')
```
